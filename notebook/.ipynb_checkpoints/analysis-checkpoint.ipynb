# Parkee POS — Analisis Penjualan

Koneksi langsung ke ClickHouse (mart layer) untuk menjawab 3 pertanyaan analitik:

1. **Q1** — 5 produk terlaris per kategori (berdasarkan total quantity terjual)
2. **Q2** — Trend total revenue per bulan
3. **Q3** — Persentase penggunaan tiap metode pembayaran

```python
import os

import clickhouse_connect
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

client = clickhouse_connect.get_client(
    host=os.environ.get("CLICKHOUSE_HOST", "localhost"),
    port=int(os.environ.get("CLICKHOUSE_HTTP_PORT", "8123")),
    username=os.environ.get("CLICKHOUSE_USER", "parkee"),
    password=os.environ.get("CLICKHOUSE_PASSWORD", "parkee"),
)
client.query("select 1").result_rows
```

## Q1 — 5 Produk Terlaris per Kategori (Total Quantity Terjual)

```python
q1_sql = """
select category, product_name, total_quantity
from (
    select
        p.category as category,
        p.product_name as product_name,
        sum(f.quantity) as total_quantity,
        row_number() over (partition by p.category order by sum(f.quantity) desc) as rn
    from analytics_marts.fact_sales as f
    inner join analytics_marts.dim_product as p on f.product_id = p.product_id
    group by p.category, p.product_name
)
where rn <= 5
order by category, total_quantity desc
"""

df_q1 = client.query_df(q1_sql)
df_q1
```

```python
categories = df_q1["category"].unique()
n_cat = len(categories)

fig, axes = plt.subplots(n_cat, 1, figsize=(9, 3.2 * n_cat))
if n_cat == 1:
    axes = [axes]

for ax, category in zip(axes, categories):
    subset = df_q1[df_q1["category"] == category].sort_values("total_quantity")
    ax.barh(subset["product_name"], subset["total_quantity"], color="#4C72B0")
    ax.set_title(f"Top 5 Produk Terlaris — {category}")
    ax.set_xlabel("Total Quantity Terjual")
    ax.set_ylabel("")

plt.tight_layout()
plt.show()
```

## Q2 — Trend Total Revenue per Bulan

```python
q2_sql = """
select
    toStartOfMonth(transaction_date) as month,
    sum(subtotal) as revenue
from analytics_marts.fact_sales
group by month
order by month
"""

df_q2 = client.query_df(q2_sql)
df_q2["month"] = pd.to_datetime(df_q2["month"])
df_q2
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df_q2["month"], df_q2["revenue"], marker="o", color="#DD8452")
ax.set_title("Trend Total Revenue per Bulan")
ax.set_xlabel("Bulan")
ax.set_ylabel("Total Revenue")
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

## Q3 — Persentase Penggunaan Tiap Metode Pembayaran

```python
q3_sql = """
select
    payment_method,
    count(distinct transaction_id) as transaction_count
from analytics_marts.fact_sales
group by payment_method
order by transaction_count desc
"""

df_q3 = client.query_df(q3_sql)
df_q3["pct"] = (df_q3["transaction_count"] / df_q3["transaction_count"].sum() * 100).round(2)
df_q3
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
colors = sns.color_palette("Set2", len(df_q3))
wedges, _, autotexts = ax.pie(
    df_q3["transaction_count"],
    labels=df_q3["payment_method"],
    autopct="%1.1f%%",
    colors=colors,
    wedgeprops={"width": 0.45, "edgecolor": "white"},
    pctdistance=0.78,
)
ax.set_title("Persentase Penggunaan Metode Pembayaran")
plt.tight_layout()
plt.show()
```
